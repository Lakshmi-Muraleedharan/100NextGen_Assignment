```python
#Part1
loader=PyPDFLoader("/content/troubleshooting_runbook.pdf")
```

```python
documents=loader.load()
report=""
for page in documents:
  report+=page.page_content+'\n'
print("Total no of pages",len(documents))
print(report[:500])
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
text_splitter=RecursiveCharacterTextSplitter(
    chunk_size=800,
    chunk_overlap=100,
)
chunk=text_splitter.split_documents(documents)
print(len(chunk))
```

```python
#Part2
from  langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_community.vectorstores import FAISS
```

```python
import os
import getpass
```

```python
os.environ["GEMINI_API_KEY"]=getpass.getpass("Enter API Key")
```

```python
embedding=GoogleGenerativeAIEmbeddings(
    model="gemini-embedding-001" # Converts text chunk into numeric vectors,
                                  #so chunks with similar meaning can be stay together in vector space
)
```

```python
vector_store=FAISS.from_documents( #FAISS as it keeps all vectis in a in-memory index and is queried by distance.
    documents=chunk,                #Each chunk keeps it page number, so ans can cite pages.
    embedding=embedding
)
```

```python
query="What policy does API gateway use to enforce ratelimiting"
for k in[3,4,7]:
  result=vector_store.similarity_search_with_score(query,k=k)
  print(f"----------------------")
  for i,(doc,score)in enumerate (result):
    print(f"Result{i}")
    print(f"Document:",doc.page_content[:500])
    print(f"Result Score:{score}")
```

```python
retriver=vector_store.as_retriever(
    search_type="similarity",
    search_kwargs={"k":2}
)
```

```python
#Part 4
from langchain_core.prompts import ChatPromptTemplate
```

```python
#Prompt design
#The context is restricted to retrived content and the model is instructed to not assume or use general informantion.
#If the information is not available, say Info not available insteas of guessing and while providing response give evidence.
prompt=ChatPromptTemplate.from_template(
    """
    You are a Q&A assistant who answers questions on troubleshooting runbook.
    Please provide response based on following rules:
    - Use only the data provided below to generate the response
    - Donot assume or use general information to generate response
    - If the answer cannot be found in the provided playbook, state that the information is not available in
the document rather than generating an answer from general knowledge
    - Keep the answer consise and complete
    - If the data contains page reference, include it in the response
    context:{context}
    question:{question}
    Provode the following:
    1. Answer
    2. Supporting evidence
    """

)
```

```python
from langchain_core.output_parsers import StrOutputParser
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm=ChatGoogleGenerativeAI(model='gemini-3.6-flash')
chain=prompt|llm|StrOutputParser()
```

```python
question="What policy does API gateway use to enforce ratelimiting"
```

```python
retrived_docs=retriver.invoke(question)# Example of retrieved chunks for sample query
context="\n\n".join(
    [doc.page_content for doc in retrived_docs])
response=chain.invoke({"context":context,"question":question})
print(response)
```

```python
#Part 5
while True:
    question = input('Ask a question: ').strip()
    if question.lower() in ('quit', 'exit', ''):
        break
    retrived_docs=retriver.invoke(question)
    context="\n\n".join(
        [doc.page_content for doc in retrived_docs])
    response=chain.invoke({"context":context,"question":question})
    print(response)
```
